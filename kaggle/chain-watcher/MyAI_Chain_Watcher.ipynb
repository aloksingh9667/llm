# MyAI Chain Watcher (CPU)

Scheduled by Kaggle (no local terminal needed). Each run:

1. Mints a fresh 12h API token from the private `myai-kaggle-creds` dataset
2. Scans the leg chain (`myai-100m-train-leg1..20`) via the API
3. Repairs what is broken:
   - a leg pinned **QUEUED** too long → version-bump re-push (twice, 55 min apart)
   - a leg **ERROR/CANCELLED** → re-push (deterministic resume, no progress lost)
   - a **missing next leg** after a completed one → re-creates it (self-heal of a
     failed kernel-side chain push)
   - all COMPLETE → prints chain-done and exits

CPU-only: never consumes GPU quota.
